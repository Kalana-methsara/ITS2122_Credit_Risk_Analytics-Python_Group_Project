# Target Variable Distribution Analysis

* Visualize the overall default rate using a clear side-by-side **Pie Chart** and **Count Plot**.
* Calculate and display the exact percentage of customers who defaulted.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

default_counts = df['default_next_month'].value_counts()
axes[0].pie(default_counts, labels=['Non-Default (0)', 'Default (1)'], 
            autopct='%1.1f%%', colors=["#698ddb", "#c01818"], startangle=90)
axes[0].set_title('Overall Default Rate Distribution', fontsize=14, pad=15)

ax2 = sns.countplot(x='default_next_month', data=df, palette=["#698ddb", "#c01818"], ax=axes[1])
axes[1].set_title('Distribution of Target Variable (Default Next Month)', fontsize=14, pad=15)
axes[1].set_xlabel('Default Status (0 = No, 1 = Yes)', fontsize=12)
axes[1].set_ylabel('Number of Customers', fontsize=12)
axes[1].set_xticklabels(['Non-Defaulter (0)', 'Defaulter (1)'])

total = len(df)
for p in ax2.patches:
    height = p.get_height()
    percentage = f'{(height / total) * 100:.1f}%'
    ax2.annotate(percentage,
                 (p.get_x() + p.get_width() / 2., height),
                 ha='center', va='center',
                 xytext=(0, 5),
                 textcoords='offset points',
                 fontweight='bold', fontsize=11)

plt.tight_layout()
plt.show()

print(f"Overall Default Rate: {(df['default_next_month'].mean() * 100):.2f}%")
```

# Demographic Impact on Default Risk

* Analyze and compare default propensities across multiple demographic factors using a multi-panel subplot grid.
* Visualize relationships for **Gender**, **Education Level**, **Age Group**, and **Marital Status** with corresponding data labels.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# gender

ax1 = sns.countplot(x='gender', hue='default_next_month', data=df, palette=["#4e83d3", "#e260fc"], ax=axes[0, 0])
axes[0, 0].set_title('Default Propensity by Gender', fontsize=13, pad=12)
axes[0, 0].set_xlabel('Gender', fontsize=11)
axes[0, 0].set_ylabel('Number of Customers', fontsize=11)
axes[0, 0].legend(title='Default Status', labels=['Non-Defaulter (0)', 'Defaulter (1)'])

total_n = len(df)
for p in ax1.patches:
    height = p.get_height()
    if height > 0:
        percentage = f'{(height / total_n) * 100:.1f}%'
        ax1.annotate(percentage,
                     (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='center',
                     xytext=(0, 5), textcoords='offset points',
                     fontsize=9, fontweight='bold')

# educational level

ax2 = sns.countplot(x='education', hue='default_next_month', data=df, palette=["#4e83d3", "#e260fc"], ax=axes[0, 1])
axes[0, 1].set_title('Default Propensity by Education Level', fontsize=13, pad=12)
axes[0, 1].set_xlabel('Education Level', fontsize=11)
axes[0, 1].set_ylabel('Number of Customers', fontsize=11)
axes[0, 1].legend(title='Default Status', labels=['Non-Defaulter (0)', 'Defaulter (1)'])
axes[0, 1].tick_params(axis='x', rotation=15)

for p in ax2.patches:
    height = p.get_height()
    if height > 0:
        percentage = f'{(height / total_n) * 100:.1f}%'
        ax2.annotate(percentage,
                     (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='center',
                     xytext=(0, 5), textcoords='offset points',
                     fontsize=9, fontweight='bold')

# age group

ax3 = sns.countplot(x='age_group', hue='default_next_month', data=df, palette=["#4e83d3", "#e260fc"], ax=axes[1, 0])
axes[1, 0].set_title('Default Propensity by Age Group', fontsize=13, pad=12)
axes[1, 0].set_xlabel('Age Group', fontsize=11)
axes[1, 0].set_ylabel('Number of Customers', fontsize=11)
axes[1, 0].legend(title='Default Status', labels=['Non-Defaulter (0)', 'Defaulter (1)'])
axes[1, 0].tick_params(axis='x', rotation=15)

for p in ax3.patches:
    height = p.get_height()
    if height > 0:
        percentage = f'{(height / total_n) * 100:.1f}%'
        ax3.annotate(percentage,
                     (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='center',
                     xytext=(0, 5), textcoords='offset points',
                     fontsize=9, fontweight='bold')

# marital-status

ax4 = sns.countplot(x='marital_status', hue='default_next_month', data=df, palette=["#4e83d3", "#e260fc"], ax=axes[1, 1])
axes[1, 1].set_title('Default Propensity by Marital Status', fontsize=13, pad=12)
axes[1, 1].set_xlabel('Marital Status', fontsize=11)
axes[1, 1].set_ylabel('Number of Customers', fontsize=11)
axes[1, 1].legend(title='Default Status', labels=['Non-Defaulter (0)', 'Defaulter (1)'])

for p in ax4.patches:
    height = p.get_height()
    if height > 0:
        percentage = f'{(height / total_n) * 100:.1f}%'
        ax4.annotate(percentage,
                     (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='center',
                     xytext=(0, 5), textcoords='offset points',
                     fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

# Correlation Matrix Analysis

* Construct a correlation heatmap to analyze the linear relationships between key financial metrics and default risk.
* Evaluate multicollinearity and structural patterns among numerical variables like credit limits, bill amounts, and repayment statuses.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="white")

plt.figure(figsize=(10, 8))

corr_cols = ['limit_balance', 'age', 'total_bill_amount', 'pay_amt_1', 'pay_amt_2', 'pay_amt_3', 'pay_status_1', 'pay_status_2', 'default_next_month']
correlation = df[corr_cols].corr()

sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', vmin=-0.3, vmax=1.0, 
            linewidths=0.5, cbar_kws={"shrink": 0.8}, annot_kws={"size": 10})

plt.title('Correlation Matrix of Key Financial & Risk Variables', fontsize=14, pad=15)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)

plt.tight_layout()
plt.show()
```

# Advanced Financial & Behavioral Risk Visualizations

* Generate a multi-panel subplot grid to analyze advanced risk factors:
  * **Viz 5:** Credit limit balance distributions across default states.
  * **Viz 6:** Default frequencies segmented by age groups.
  * **Viz 7:** Impact of total delayed payment months on default risk.
  * **Viz 8:** Scatter plot examining average bill amounts versus average payment amounts (zoomed view).

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

sns.boxplot(x='default_next_month', y='limit_balance', data=df, palette=["#178d32", "#e47636"], ax=axes[0, 0])
axes[0, 0].set_title('Viz 5: Credit Limit Balance by Default Status', fontsize=12, pad=12)
axes[0, 0].set_xlabel('default_next_month', fontsize=10)
axes[0, 0].set_ylabel('limit_balance', fontsize=10)
axes[0, 0].set_xticklabels(['Non-Defaulter (0)', 'Defaulter (1)'])

sns.countplot(x='age_group', hue='default_next_month', data=df, palette=["#f05949", "#4341c9"], ax=axes[0, 1])
axes[0, 1].set_title('Viz 6: Default Frequency across Age Groups', fontsize=12, pad=12)
axes[0, 1].set_xlabel('age_group', fontsize=10)
axes[0, 1].set_ylabel('count', fontsize=10)
axes[0, 1].legend(title='Default', labels=['No', 'Yes'])
axes[0, 1].tick_params(axis='x', rotation=15)

sns.countplot(x='delayed_months_count', hue='default_next_month', data=df, palette=["#16ce44", '#f1c40f'], ax=axes[1, 0])
axes[1, 0].set_title('Viz 7: Impact of Total Delayed Months on Default Risk', fontsize=12, pad=12)
axes[1, 0].set_xlabel('delayed_months_count', fontsize=10)
axes[1, 0].set_ylabel('count', fontsize=10)
axes[1, 0].legend(title='Default', labels=['No', 'Yes'])

sns.scatterplot(x='avg_bill_amt', y='avg_pay_amt', hue='default_next_month', data=df, 
                palette=['#3498db', '#e67e22'], alpha=0.6, s=25, ax=axes[1, 1])
axes[1, 1].set_title('Viz 8: Average Bill Amount vs. Average Payment Amount (Zoomed)', fontsize=12, pad=12)
axes[1, 1].set_xlabel('avg_bill_amt', fontsize=10)
axes[1, 1].set_ylabel('avg_pay_amt', fontsize=10)
axes[1, 1].legend(title='default_next_month', labels=['0', '1'])

axes[1, 1].set_ylim(0, 50000)

plt.tight_layout()
plt.show()
```
